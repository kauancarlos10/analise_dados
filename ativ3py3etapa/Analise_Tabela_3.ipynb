# Análise da Tabela 3

## Objetivo
Dividir a Tabela 3 em três tabelas por região, separando homens e mulheres, e apresentar as estatísticas descritivas usando `describe()`.

**Fonte dos dados:** IBGE – Censo Demográfico.

```python
import pandas as pd
arquivo = 'Tabela_3_original.xlsx'
df = pd.read_excel(arquivo, header=None)
df.head()
```

```python
dados = df.iloc[3:36, :10].copy()
dados.columns = [
    'Região/UF', 'Superior_Total', 'Superior_Homens', 'Superior_Mulheres',
    'CTEM_Total', 'CTEM_Homens', 'CTEM_Mulheres',
    'Edu_Saude_BemEstar_Total', 'Edu_Saude_BemEstar_Homens',
    'Edu_Saude_BemEstar_Mulheres'
]

for c in dados.columns[1:]:
    dados[c] = pd.to_numeric(dados[c], errors='coerce')

grandes_regioes = ['Brasil', 'Norte', 'Nordeste', 'Sudeste', 'Sul', 'Centro-Oeste']
regioes = dados[dados['Região/UF'].isin(grandes_regioes)].reset_index(drop=True)
regioes
```

## 2. Tabela 1 – Homens e mulheres por região com curso superior

```python
tabela1 = regioes[['Região/UF', 'Superior_Total', 'Superior_Homens', 'Superior_Mulheres']].copy()
tabela1.columns = ['Região', 'Total', 'Homens', 'Mulheres']
tabela1
```

```python
tabela1[['Total', 'Homens', 'Mulheres']].describe()
```

```python
tabela2 = regioes[['Região/UF', 'CTEM_Total', 'CTEM_Homens', 'CTEM_Mulheres']].copy()
tabela2.columns = ['Região', 'Total', 'Homens', 'Mulheres']
tabela2
```

```python
tabela2[['Total', 'Homens', 'Mulheres']].describe()
```

```python
tabela3 = regioes[['Região/UF', 'Edu_Saude_BemEstar_Total', 'Edu_Saude_BemEstar_Homens', 'Edu_Saude_BemEstar_Mulheres']].copy()
tabela3.columns = ['Região', 'Total', 'Homens', 'Mulheres']
tabela3
```

```python
tabela3[['Total', 'Homens', 'Mulheres']].describe()
```

```python
tabela1.to_excel('01_curso_superior_por_regiao.xlsx', index=False)
tabela2.to_excel('02_CTEM_por_regiao.xlsx', index=False)
tabela3.to_excel('03_Educacao_Servicos_Saude_Bem_Estar_por_regiao.xlsx', index=False)
print('Arquivos salvos com sucesso!')
```
